```python
import duckdb
import polars as pl

def stream_remote_amazon_subset(limit_products: int = 10000):
    con = duckdb.connect(database=":memory:")
    
    # Load DuckDB's HTTP/HuggingFace file system extension
    con.execute("INSTALL httpfs; LOAD httpfs;")
    
    # Stream directly from the Hugging Face repository without saving raw files locally
    # Example using the All_Beauty or Clothing subset hosted on Hugging Face
    remote_meta_url = (
        "https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/"
        "resolve/main/raw/meta_categories/meta_All_Beauty.jsonl"
    )

    query = f"""
    SELECT
        parent_asin,
        TRIM(title) AS title,
        TRY_CAST(REGEXP_EXTRACT(CAST(price AS VARCHAR), '([0-9]+\\.?[0-9]*)', 1) AS DOUBLE) AS clean_price,
        CAST(average_rating AS DOUBLE) AS avg_rating,
        CAST(rating_number AS BIGINT) AS rating_count,
        COALESCE(images[1].hi_res, images[1].large) AS image_url
    FROM read_json_auto('{remote_meta_url}', ignore_errors=true)
    WHERE price IS NOT NULL
      AND TRY_CAST(REGEXP_EXTRACT(CAST(price AS VARCHAR), '([0-9]+\\.?[0-9]*)', 1) AS DOUBLE) > 0
      AND LEN(images) > 0
      AND COALESCE(images[1].hi_res, images[1].large) IS NOT NULL
    LIMIT {limit_products};
    """
    
    # Materialize only the tiny filtered table (~2-5 MB) into Polars
    df = con.execute(query).pl()
    con.close()
    return df
```

```python
import duckdb
import polars as pl
from pathlib import Path

def build_tabular_dataset_remotely(
    category: str = "All_Beauty",  # Can be changed to 'Clothing_Shoes_and_Jewelry', 'Home_and_Kitchen', etc.
    output_dir: str = "data/processed",
    max_products: int = 15000
):
    """
    Streams UCSD Amazon 2023 metadata and review logs directly from Hugging Face 
    into RAM via DuckDB httpfs, engineers temporal features in Polars, and exports 
    compact Parquet tables (~15-30 MB total). Zero raw dataset files are downloaded.
    """
    Path(output_dir).mkdir(parents=True, exist_ok=True)
    con = duckdb.connect(database=":memory:")
    
    # Configure in-memory limits and install HTTP/HuggingFace file system support
    con.execute("SET memory_limit = '6GB';")
    con.execute("INSTALL httpfs; LOAD httpfs;")
    con.execute("SET enable_http_metadata_cache = true;")

    # Remote Hugging Face URLs (streaming directly over HTTPS)
    meta_remote_url = (
        f"https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/"
        f"resolve/main/raw/meta_categories/meta_{category}.jsonl"
    )
    reviews_remote_url = (
        f"https://huggingface.co/datasets/McAuley-Lab/Amazon-Reviews-2023/"
        f"resolve/main/raw/review_categories/{category}.jsonl"
    )

    print(f"1. Streaming and cleaning '{category}' metadata directly from Hugging Face...")
    meta_query = f"""
    CREATE TEMP TABLE clean_meta AS
    WITH raw_parsed AS (
        SELECT
            parent_asin,
            TRIM(title) AS title,
            main_category,
            TRY_CAST(
                REGEXP_EXTRACT(CAST(price AS VARCHAR), '([0-9]+\\.?[0-9]*)', 1) 
                AS DOUBLE
            ) AS clean_price,
            CAST(average_rating AS DOUBLE) AS catalog_avg_rating,
            CAST(rating_number AS BIGINT) AS total_catalog_ratings,
            LEN(features) AS bullet_count,
            COALESCE(
                list_extract(list_filter(images, x -> x.variant = 'MAIN'), 1).hi_res,
                list_extract(list_filter(images, x -> x.variant = 'MAIN'), 1).large,
                images[1].hi_res,
                images[1].large
            ) AS image_url,
            json_extract_string(details, '$.Brand') AS brand
        FROM read_json_auto(
            '{meta_remote_url}',
            format='newline_delimited',
            ignore_errors=true,
            maximum_object_size=33554432
        )
    )
    SELECT *
    FROM raw_parsed
    WHERE clean_price IS NOT NULL
      AND clean_price BETWEEN 1.0 AND 2500.0
      AND image_url IS NOT NULL
      AND STARTS_WITH(image_url, 'http')
      AND total_catalog_ratings >= 5
    LIMIT {max_products};
    """
    con.execute(meta_query)

    # Export clean dimension table (~2-5 MB)
    dim_products_pl = con.execute("SELECT * FROM clean_meta").pl()
    dim_products_pl.write_parquet(f"{output_dir}/dim_products.parquet", compression="zstd")
    print(f"   -> Cached {len(dim_products_pl)} clean products to dim_products.parquet")

    print("2. Streaming remote review logs and aggregating monthly demand in RAM...")
    reviews_query = f"""
    CREATE TEMP TABLE monthly_reviews AS
    SELECT
        r.parent_asin,
        DATE_TRUNC('month', to_timestamp(r.timestamp / 1000.0))::DATE AS review_month,
        COUNT(*) AS monthly_review_count,
        AVG(CAST(r.rating AS DOUBLE)) AS monthly_avg_rating,
        SUM(CASE WHEN r.verified_purchase THEN 1 ELSE 0 END) AS verified_count
    FROM read_json_auto(
        '{reviews_remote_url}',
        format='newline_delimited',
        ignore_errors=true
    ) r
    INNER JOIN clean_meta m ON r.parent_asin = m.parent_asin
    WHERE r.timestamp IS NOT NULL
    GROUP BY r.parent_asin, review_month;
    """
    con.execute(reviews_query)

    # Hand off aggregated monthly table to Polars via zero-copy Arrow
    fact_df = con.execute("SELECT * FROM monthly_reviews ORDER BY parent_asin, review_month").pl()
    con.close()  # Immediately frees all DuckDB in-memory tables

    print("3. Computing temporal lags and 30-day forward demand target in Polars...")
    fact_engineered = (
        fact_df
        .sort(["parent_asin", "review_month"])
        .with_columns([
            # Historical lag features (Strictly t-1 and t-2 to prevent time leakage)
            pl.col("monthly_review_count").shift(1).over("parent_asin").fill_null(0).alias("reviews_lag_1m"),
            pl.col("monthly_review_count").shift(2).over("parent_asin").fill_null(0).alias("reviews_lag_2m"),
            
            # 3-month rolling historical velocity
            pl.col("monthly_review_count")
              .rolling_mean(window_size=3, min_samples=1)
              .over("parent_asin")
              .alias("rolling_3m_velocity"),

            # Verified purchase ratio
            (pl.col("verified_count") / pl.col("monthly_review_count")).alias("verified_ratio"),

            # TARGET: Next active month's review velocity
            pl.col("monthly_review_count").shift(-1).over("parent_asin").alias("target_next_month_reviews")
        ])
        .filter(pl.col("target_next_month_reviews").is_not_null())
        .with_columns([
            # Log1p target transformation: y = log(1 + Reviews_{t+1})
            pl.col("target_next_month_reviews").log1p().alias("target_log_demand")
        ])
    )

    fact_engineered.write_parquet(f"{output_dir}/fact_monthly_demand.parquet", compression="zstd")
    print(f"   -> Saved {len(fact_engineered)} product-month rows to fact_monthly_demand.parquet")

if __name__ == "__main__":
    build_tabular_dataset_remotely(
        category="All_Beauty",  # Starts fast (~700k reviews); switch to larger categories anytime
        output_dir="data/processed",
        max_products=15000
    )
```

```python
import asyncio
import aiohttp
import io
import cv2
import torch
import numpy as np
import polars as pl
import imagehash
from PIL import Image, UnidentifiedImageError
from torchvision import models
from tqdm.asyncio import tqdm
from dataclasses import dataclass, asdict
from typing import Optional, List, Tuple
from pathlib import Path

# -------------------------------------------------------------------------
# 1. Data Structure for Lightweight Output (Zero JPEG storage)
# -------------------------------------------------------------------------
@dataclass
class InMemoryImageRecord:
    parent_asin: str
    status: str  # 'VALID', 'CORRUPT', 'LOW_RES', 'BLANK_PLACEHOLDER', 'HTTP_ERROR'
    width: Optional[int] = None
    height: Optional[int] = None
    aspect_ratio: Optional[float] = None
    sharpness_score: Optional[float] = None
    brightness_mean: Optional[float] = None
    contrast_std: Optional[float] = None
    white_bg_ratio: Optional[float] = None
    phash: Optional[str] = None
    vision_embedding: Optional[List[float]] = None  # 576-d vector (~1.1 KB in Parquet)


# -------------------------------------------------------------------------
# 2. Initialize Pretrained Vision Backbone in Memory
# -------------------------------------------------------------------------
def load_feature_extractor(device: str) -> Tuple[torch.nn.Module, callable]:
    weights = models.MobileNet_V3_Small_Weights.DEFAULT
    backbone = models.mobilenet_v3_small(weights=weights).to(device).eval()
    # Remove classification head -> outputs a 576-dimensional latent vector
    extractor = torch.nn.Sequential(
        backbone.features,
        backbone.avgpool,
        torch.nn.Flatten()
    ).to(device).eval()
    return extractor, weights.transforms()


# -------------------------------------------------------------------------
# 3. CPU Worker: In-Memory Hygiene Audit, OpenCV Metrics & Tensor Prep
# -------------------------------------------------------------------------
def audit_and_preprocess_in_ram(
    raw_bytes: bytes,
    parent_asin: str,
    preprocess_fn: callable,
    min_dim: int = 128
) -> Tuple[InMemoryImageRecord, Optional[torch.Tensor]]:
    """
    Decodes bytes in RAM, runs OpenCV quality checks, and returns a normalized
    224x224 PyTorch tensor. Raw bytes and PIL objects are freed upon exit.
    """
    try:
        pil_img = Image.open(io.BytesIO(raw_bytes))
        pil_img.verify()
        pil_img = Image.open(io.BytesIO(raw_bytes)).convert("RGB")
    except (UnidentifiedImageError, OSError, ValueError):
        return InMemoryImageRecord(parent_asin=parent_asin, status="CORRUPT"), None

    width, height = pil_img.size
    if width < min_dim or height < min_dim:
        return InMemoryImageRecord(
            parent_asin=parent_asin, status="LOW_RES", width=width, height=height
        ), None

    rgb_arr = np.array(pil_img)
    gray_arr = cv2.cvtColor(rgb_arr, cv2.COLOR_RGB2GRAY)

    # 1. Blank / solid placeholder filter
    brightness_mean = float(np.mean(gray_arr))
    contrast_std = float(np.std(gray_arr))
    if contrast_std < 3.0 or brightness_mean > 253.0 or brightness_mean < 2.0:
        return InMemoryImageRecord(
            parent_asin=parent_asin,
            status="BLANK_PLACEHOLDER",
            width=width,
            height=height,
            brightness_mean=round(brightness_mean, 2),
            contrast_std=round(contrast_std, 2)
        ), None

    # 2. Laplacian Sharpness: Var(\nabla^2 I)
    sharpness_score = float(cv2.Laplacian(gray_arr, cv2.CV_64F).var())

    # 3. Border Whiteness Ratio (studio background detection)
    b_w = max(1, int(min(width, height) * 0.05))
    border_pixels = np.concatenate([
        gray_arr[:b_w, :].ravel(),
        gray_arr[-b_w:, :].ravel(),
        gray_arr[:, :b_w].ravel(),
        gray_arr[:, -b_w:].ravel()
    ])
    white_bg_ratio = float(np.mean(border_pixels > 240))

    # 4. 64-bit Perceptual Hash for duplicate image tracking
    phash_str = str(imagehash.phash(pil_img))

    # 5. Transform PIL image into normalized (3, 224, 224) tensor in RAM
    img_tensor = preprocess_fn(pil_img)

    record = InMemoryImageRecord(
        parent_asin=parent_asin,
        status="VALID",
        width=width,
        height=height,
        aspect_ratio=round(width / height, 4),
        sharpness_score=round(sharpness_score, 2),
        brightness_mean=round(brightness_mean, 2),
        contrast_std=round(contrast_std, 2),
        white_bg_ratio=round(white_bg_ratio, 4),
        phash=phash_str
    )
    return record, img_tensor


# -------------------------------------------------------------------------
# 4. Async Network Fetcher
# -------------------------------------------------------------------------
async def fetch_and_audit_one(
    session: aiohttp.ClientSession,
    semaphore: asyncio.Semaphore,
    parent_asin: str,
    url: str,
    preprocess_fn: callable
) -> Tuple[InMemoryImageRecord, Optional[torch.Tensor]]:
    async with semaphore:
        for attempt in range(3):
            try:
                async with session.get(url, timeout=aiohttp.ClientTimeout(total=10)) as resp:
                    if resp.status == 200:
                        raw_bytes = await resp.read()
                        return await asyncio.to_thread(
                            audit_and_preprocess_in_ram,
                            raw_bytes,
                            parent_asin,
                            preprocess_fn
                        )
                    elif resp.status == 404:
                        return InMemoryImageRecord(parent_asin=parent_asin, status="HTTP_404"), None
            except (aiohttp.ClientError, asyncio.TimeoutError):
                if attempt < 2:
                    await asyncio.sleep(0.4 * (2 ** attempt))
                continue
        return InMemoryImageRecord(parent_asin=parent_asin, status="HTTP_ERROR"), None


# -------------------------------------------------------------------------
# 5. Batched Stream Runner (Bounds RAM and Accelerates GPU Inference)
# -------------------------------------------------------------------------
async def run_zero_disk_vision_pipeline(
    dim_products_parquet: str = "data/processed/dim_products.parquet",
    output_metadata_parquet: str = "data/processed/dim_image_features.parquet",
    chunk_size: int = 512,
    concurrency: int = 64,
    inference_batch_size: int = 64
):
    Path(output_metadata_parquet).parent.mkdir(parents=True, exist_ok=True)
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Initializing in-memory vision backbone on [{device.upper()}]...")
    extractor, preprocess_fn = load_feature_extractor(device)

    df = pl.read_parquet(dim_products_parquet).select(["parent_asin", "image_url"])
    rows = df.to_dicts()
    total_rows = len(rows)

    semaphore = asyncio.Semaphore(concurrency)
    connector = aiohttp.TCPConnector(limit=concurrency, ttl_dns_cache=300)
    headers = {"User-Agent": "Mozilla/5.0 (compatible; ZeroDiskAnalytics/1.0)"}

    all_completed_records: List[dict] = []

    async with aiohttp.ClientSession(connector=connector, headers=headers) as session:
        # Process in chunks of `chunk_size` so RAM never accumulates thousands of tensors
        for start_idx in tqdm(range(0, total_rows, chunk_size), desc="Streaming Chunks"):
            chunk = rows[start_idx : start_idx + chunk_size]
            tasks = [
                fetch_and_audit_one(session, semaphore, r["parent_asin"], r["image_url"], preprocess_fn)
                for r in chunk
            ]
            chunk_results = await asyncio.gather(*tasks)

            # Separate valid records that have tensors from failed/filtered records
            valid_records: List[InMemoryImageRecord] = []
            valid_tensors: List[torch.Tensor] = []

            for record, tensor in chunk_results:
                if tensor is not None and record.status == "VALID":
                    valid_records.append(record)
                    valid_tensors.append(tensor)
                else:
                    all_completed_records.append(asdict(record))

            # Run batched neural network inference on valid tensors in this chunk
            if valid_tensors:
                with torch.no_grad():
                    for b_start in range(0, len(valid_tensors), inference_batch_size):
                        batch_slice = valid_tensors[b_start : b_start + inference_batch_size]
                        batch_tensor = torch.stack(batch_slice).to(device)
                        
                        # Extract 576-d embedding and L2-normalize for cosine similarity
                        feats = extractor(batch_tensor)
                        feats = torch.nn.functional.normalize(feats, p=2, dim=1)
                        feats_np = feats.cpu().numpy().astype(np.float32)

                        for idx, vec in enumerate(feats_np):
                            rec = valid_records[b_start + idx]
                            rec.vision_embedding = np.round(vec, 5).tolist()
                            all_completed_records.append(asdict(rec))

                # Explicitly release tensor list from RAM before next chunk
                del valid_tensors

    # Save final compact table with Polars Array/List column
    features_df = pl.DataFrame(all_completed_records)

    # Cast embedding list to Float32 for 50% smaller Parquet footprint
    features_df = features_df.with_columns([
        pl.col("vision_embedding").cast(pl.List(pl.Float32)),
        pl.when(pl.col("phash").is_not_null())
          .then(pl.col("phash").count().over("phash") > 1)
          .otherwise(False)
          .alias("is_duplicate_stock_photo")
    ])

    features_df.write_parquet(output_metadata_parquet, compression="zstd")
    print(f"\nSaved {len(features_df)} records to {output_metadata_parquet} (0 JPEGs written to disk).")
    print(features_df.group_by("status").len().sort("len", descending=True))

if __name__ == "__main__":
    await run_zero_disk_vision_pipeline()
```

```python
import polars as pl
import numpy as np
from scipy import stats
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# -------------------------------------------------------------------------
# 1. Load & Merge the Zero-Disk Parquet Feature Store
# -------------------------------------------------------------------------
dim_prod = pl.read_parquet("data/processed/dim_products.parquet")
dim_img = pl.read_parquet("data/processed/dim_image_features.parquet")
fact_demand = pl.read_parquet("data/processed/fact_monthly_demand.parquet")

# Keep only products with VALID audited images and join with demand history
valid_img = dim_img.filter(pl.col("status") == "VALID").drop("vision_embedding")

# Product-level summary (aggregating mean forward demand per unique product)
product_summary = (
    fact_demand
    .group_by("parent_asin")
    .agg([
        pl.col("target_next_month_reviews").mean().alias("mean_monthly_demand"),
        pl.col("target_log_demand").mean().alias("mean_log_demand"),
        pl.col("rolling_3m_velocity").mean().alias("avg_historical_velocity"),
        pl.col("verified_ratio").mean().alias("avg_verified_ratio"),
        pl.len().alias("active_months")
    ])
    .join(dim_prod, on="parent_asin", how="inner")
    .join(valid_img, on="parent_asin", how="inner")
    .with_columns([
        pl.col("clean_price").log1p().alias("log_price"),
        pl.col("sharpness_score").log1p().alias("log_sharpness"),
        pl.when(pl.col("white_bg_ratio") >= 0.85)
          .then(pl.lit("Studio White BG (>=85%)"))
          .otherwise(pl.lit("Lifestyle / Cluttered BG (<85%)"))
          .alias("background_type")
    ])
)

print(f"Merged Analytical Dataset: {len(product_summary)} unique validated products.")

# -------------------------------------------------------------------------
# 2. Statistical Hypothesis Testing Suite
# -------------------------------------------------------------------------
df_np = product_summary.select([
    "log_price", "mean_log_demand", "sharpness_score", 
    "brightness_mean", "contrast_std", "white_bg_ratio",
    "is_duplicate_stock_photo", "bullet_count"
]).drop_nulls()

# Test 1: Price Elasticity via Log-Log OLS Regression
slope, intercept, r_val, p_val_price, std_err = stats.linregress(
    df_np["log_price"].to_numpy(),
    df_np["mean_log_demand"].to_numpy()
)

# Test 2: Spearman Rank Correlation (Image Sharpness vs. Demand)
rho_sharp, p_val_sharp = stats.spearmanr(
    df_np["sharpness_score"].to_numpy(),
    df_np["mean_log_demand"].to_numpy()
)

# Test 3: Mann-Whitney U Test (Studio White BG vs. Cluttered BG)
studio_demand = product_summary.filter(pl.col("white_bg_ratio") >= 0.85)["mean_log_demand"].to_numpy()
cluttered_demand = product_summary.filter(pl.col("white_bg_ratio") < 0.85)["mean_log_demand"].to_numpy()
u_stat, p_val_bg = stats.mannwhitneyu(studio_demand, cluttered_demand, alternative="two-sided")

# Test 4: Duplicate Stock Photo Impact (Unique vs. Re-used Supplier Photos)
unique_photo_demand = product_summary.filter(~pl.col("is_duplicate_stock_photo"))["mean_log_demand"].to_numpy()
dup_photo_demand = product_summary.filter(pl.col("is_duplicate_stock_photo"))["mean_log_demand"].to_numpy()
if len(dup_photo_demand) > 5:
    u_dup, p_val_dup = stats.mannwhitneyu(unique_photo_demand, dup_photo_demand, alternative="two-sided")
else:
    u_dup, p_val_dup = np.nan, np.nan

# Display Clean Statistical Report Table
hypothesis_report = pl.DataFrame({
    "Hypothesis / Metric": [
        "H1: Price Elasticity (Log-Price Slope)",
        "H2: Image Sharpness Rank Correlation (Spearman rho)",
        "H3: Studio White BG vs. Cluttered BG (Mann-Whitney U)",
        "H4: Unique Photo vs. Duplicate Stock Photo"
    ],
    "Effect Size / Statistic": [
        f"Elasticity = {slope:.4f}",
        f"rho = {rho_sharp:.4f}",
        f"Median Diff = {np.median(studio_demand) - np.median(cluttered_demand):.4f}",
        f"Dup Photo Count = {len(dup_photo_demand)}"
    ],
    "p-value": [
        f"{p_val_price:.4e}",
        f"{p_val_sharp:.4e}",
        f"{p_val_bg:.4e}",
        f"{p_val_dup:.4e}" if not np.isnan(p_val_dup) else "N/A (Too few duplicates)"
    ],
    "Analytical Conclusion": [
        "Significant negative elasticity" if (p_val_price < 0.05 and slope < 0) else "Inelastic / Noise",
        "Sharpness significantly correlates with demand" if p_val_sharp < 0.05 else "No monotonic link",
        "Background style significantly shifts demand" if p_val_bg < 0.05 else "No significant BG effect",
        "Seller photo duplication impacts demand" if (not np.isnan(p_val_dup) and p_val_dup < 0.05) else "No duplicate penalty"
    ]
})
print("\n=== Phase 2 Statistical Hypothesis Report ===")
print(hypothesis_report)

# -------------------------------------------------------------------------
# 3. Interactive Multi-Modal Diagnostic Dashboard (Plotly)
# -------------------------------------------------------------------------
sample_vis = product_summary.sample(n=min(3000, len(product_summary)), seed=42).to_pandas()

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        f"1. Price Elasticity Curve (Slope = {slope:.3f})",
        "2. Demand by Background Hygiene (Studio vs. Lifestyle)",
        "3. Visual Sharpness Deciles vs. Mean Log Demand",
        "4. Temporal Seasonality & Monthly Review Velocity"
    )
)

# Panel 1: Log-Price vs Log-Demand with OLS Trendline
fig.add_trace(
    go.Scattergl(
        x=sample_vis["log_price"], y=sample_vis["mean_log_demand"],
        mode="markers", marker=dict(size=4, opacity=0.35, color="#2b5c8f"),
        name="Products"
    ),
    row=1, col=1
)
x_vals = np.linspace(sample_vis["log_price"].min(), sample_vis["log_price"].max(), 50)
fig.add_trace(
    go.Scatter(
        x=x_vals, y=intercept + slope * x_vals,
        mode="lines", line=dict(color="#d95f02", width=3),
        name="Elasticity Fit"
    ),
    row=1, col=1
)

# Panel 2: Box/Violin comparison of Studio White BG vs Cluttered BG
for bg_label, color in [("Studio White BG (>=85%)", "#1b9e77"), ("Lifestyle / Cluttered BG (<85%)", "#7570b3")]:
    sub = sample_vis[sample_vis["background_type"] == bg_label]
    fig.add_trace(
        go.Box(y=sub["mean_log_demand"], name=bg_label, marker_color=color, boxmean=True),
        row=1, col=2
    )

# Panel 3: Binned Sharpness Deciles vs Mean Demand
sharp_bins = (
    product_summary
    .with_columns(pl.col("log_sharpness").qcut(10, labels=[f"D{i}" for i in range(1, 11)], allow_duplicates=True).alias("sharpness_decile"))
    .group_by("sharpness_decile")
    .agg([
        pl.col("mean_log_demand").mean().alias("avg_demand"),
        pl.col("mean_log_demand").std().alias("std_demand"),
        pl.len().alias("n")
    ])
    .sort("sharpness_decile")
    .to_pandas()
)
fig.add_trace(
    go.Bar(
        x=sharp_bins["sharpness_decile"],
        y=sharp_bins["avg_demand"],
        marker_color="#386cb0",
        name="Mean Log Demand by Sharpness Decile"
    ),
    row=2, col=1
)

# Panel 4: Macro Monthly Demand Trend (Checking for seasonal spikes & split cutoff)
monthly_macro = (
    fact_demand
    .group_by("review_month")
    .agg(pl.col("monthly_review_count").sum().alias("total_category_reviews"))
    .sort("review_month")
    .tail(36) # Last 3 years
    .to_pandas()
)
fig.add_trace(
    go.Scatter(
        x=monthly_macro["review_month"],
        y=monthly_macro["total_category_reviews"],
        mode="lines+markers",
        line=dict(color="#e7298a", width=2),
        name="Monthly Category Volume"
    ),
    row=2, col=2
)

fig.update_layout(
    height=750, width=1100,
    title_text="Multi-Modal E-Commerce EDA: Pricing, Computer Vision Hygiene & Seasonality",
    showlegend=False,
    template="plotly_white"
)
fig.show()
```

```python
import polars as pl
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score, ndcg_score
import lightgbm as lgb
from pathlib import Path
import joblib

# Set deterministic seeds
np.random.seed(42)
torch.manual_seed(42)

# -------------------------------------------------------------------------
# 1. Load, Merge & Engineer Relative Category Features
# -------------------------------------------------------------------------
def load_and_prepare_master_table(processed_dir: str = "data/processed") -> pl.DataFrame:
    dim_prod = pl.read_parquet(f"{processed_dir}/dim_products.parquet")
    dim_img = pl.read_parquet(f"{processed_dir}/dim_image_features.parquet").filter(
        pl.col("status") == "VALID"
    )
    fact_demand = pl.read_parquet(f"{processed_dir}/fact_monthly_demand.parquet")

    # Join tables on parent_asin
    df = (
        fact_demand
        .join(dim_prod, on="parent_asin", how="inner")
        .join(dim_img, on="parent_asin", how="inner")
        .with_columns([
            pl.col("clean_price").log1p().alias("log_price"),
            pl.col("review_month").dt.month().alias("month_of_year"),
            # Cyclical month encoding for seasonality
            (2 * np.pi * pl.col("review_month").dt.month() / 12.0).sin().alias("month_sin"),
            (2 * np.pi * pl.col("review_month").dt.month() / 12.0).cos().alias("month_cos"),
        ])
    )
    return df.sort("review_month")


# -------------------------------------------------------------------------
# 2. Temporal + Perceptual Hash (phash) Disjoint Split
# -------------------------------------------------------------------------
def temporal_phash_disjoint_split(df: pl.DataFrame):
    """
    Splits chronologically at the 70th and 85th time percentiles, then purges
    any validation/test rows whose perceptual image hash (phash) already appeared
    in an earlier split.
    """
    unique_months = df.select("review_month").unique().sort("review_month")["review_month"].to_list()
    n_months = len(unique_months)
    
    train_cutoff = unique_months[int(n_months * 0.70)]
    val_cutoff = unique_months[int(n_months * 0.85)]

    train_df = df.filter(pl.col("review_month") <= train_cutoff)
    val_raw = df.filter((pl.col("review_month") > train_cutoff) & (pl.col("review_month") <= val_cutoff))
    test_raw = df.filter(pl.col("review_month") > val_cutoff)

    # Disjoint phash enforcement: prevent duplicate stock photos from leaking across splits
    train_phashes = set(train_df["phash"].drop_nulls().to_list())
    
    # For validation and test, we keep existing products' future months OR new products with unseen phashes
    # To strictly test generalization without visual duplication leakage across DIFFERENT products:
    train_asin_phash = dict(zip(train_df["parent_asin"].to_list(), train_df["phash"].to_list()))

    def filter_cross_seller_dup_leakage(split_df: pl.DataFrame, seen_phashes: set, allowed_asins: set):
        # Allow a row if it's the SAME product tracked over time, OR a brand new product with an unseen phash
        return split_df.filter(
            pl.col("parent_asin").is_in(list(allowed_asins)) | (~pl.col("phash").is_in(list(seen_phashes)))
        )

    train_asins = set(train_df["parent_asin"].to_list())
    val_df = filter_cross_seller_dup_leakage(val_raw, train_phashes, train_asins)
    
    seen_phashes_val = train_phashes.union(set(val_df["phash"].drop_nulls().to_list()))
    seen_asins_val = train_asins.union(set(val_df["parent_asin"].to_list()))
    test_df = filter_cross_seller_dup_leakage(test_raw, seen_phashes_val, seen_asins_val)

    print(f"Split Summary | Train: {len(train_df)} (<= {train_cutoff}) | "
          f"Val: {len(val_df)} (<= {val_cutoff}) | Test: {len(test_df)} (> {val_cutoff})")
    return train_df, val_df, test_df


# -------------------------------------------------------------------------
# 3. Evaluation Metric Helper (RMSE, MAE, R2, and Ranking NDCG@50)
# -------------------------------------------------------------------------
def evaluate_predictions(y_true: np.ndarray, y_pred: np.ndarray, model_name: str) -> dict:
    rmse = root_mean_squared_error(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    
    # NDCG@50: How accurately does the model identify the top 50 bestselling items?
    y_true_nonneg = np.clip(y_true, a_min=0, a_max=None).reshape(1, -1)
    y_pred_reshaped = y_pred.reshape(1, -1)
    ndcg_50 = ndcg_score(y_true_nonneg, y_pred_reshaped, k=min(50, len(y_true)))

    return {
        "Model": model_name,
        "RMSE (Log)": round(rmse, 4),
        "MAE (Log)": round(mae, 4),
        "R2 Score": round(r2, 4),
        "NDCG@50": round(ndcg_50, 4)
    }


# -------------------------------------------------------------------------
# 4. PyTorch Gated Multi-Modal Fusion Architecture
# -------------------------------------------------------------------------
class MultiModalDataset(Dataset):
    def __init__(self, X_tab: np.ndarray, X_vis: np.ndarray, y: np.ndarray):
        self.X_tab = torch.tensor(X_tab, dtype=torch.float32)
        self.X_vis = torch.tensor(X_vis, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32).unsqueeze(1)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X_tab[idx], self.X_vis[idx], self.y[idx]


class GatedMultiModalFusionNet(nn.Module):
    """
    Two-tower neural network with a learned Gated Modality Unit (GMU)
    that dynamically weights tabular vs. raw 576-d visual representations.
    """
    def __init__(self, tab_dim: int, vis_dim: int = 576, hidden_dim: int = 128):
        super().__init__()
        self.tab_branch = nn.Sequential(
            nn.Linear(tab_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU()
        )
        self.vis_branch = nn.Sequential(
            nn.Linear(vis_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(0.3),
            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU()
        )
        # Gating mechanism: learns weight g in [0, 1]^hidden_dim
        self.gate_layer = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.Sigmoid()
        )
        self.regressor = nn.Sequential(
            nn.Linear(hidden_dim, 64),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(64, 1)
        )

    def forward(self, x_tab: torch.Tensor, x_vis: torch.Tensor) -> torch.Tensor:
        h_tab = self.tab_branch(x_tab)
        h_vis = self.vis_branch(x_vis)
        
        gate = self.gate_layer(torch.cat([h_tab, h_vis], dim=1))
        z_fused = gate * h_tab + (1.0 - gate) * h_vis
        return self.regressor(z_fused)


def train_pytorch_fusion(
    X_tab_tr, X_vis_tr, y_tr,
    X_tab_val, X_vis_val, y_val,
    X_tab_te, X_vis_te,
    epochs: int = 20,
    batch_size: int = 256,
    lr: float = 1e-3
) -> np.ndarray:
    device = "cuda" if torch.cuda.is_available() else "cpu"
    train_loader = DataLoader(MultiModalDataset(X_tab_tr, X_vis_tr, y_tr), batch_size=batch_size, shuffle=True)
    
    model = GatedMultiModalFusionNet(tab_dim=X_tab_tr.shape[1], vis_dim=X_vis_tr.shape[1]).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    criterion = nn.HuberLoss(delta=1.0)  # Robust to extreme demand spikes

    best_val_loss = float("inf")
    best_weights = None

    X_tab_val_t = torch.tensor(X_tab_val, dtype=torch.float32).to(device)
    X_vis_val_t = torch.tensor(X_vis_val, dtype=torch.float32).to(device)
    y_val_t = torch.tensor(y_val, dtype=torch.float32).unsqueeze(1).to(device)

    for epoch in range(epochs):
        model.train()
        for bx_tab, bx_vis, by in train_loader:
            bx_tab, bx_vis, by = bx_tab.to(device), bx_vis.to(device), by.to(device)
            optimizer.zero_grad()
            preds = model(bx_tab, bx_vis)
            loss = criterion(preds, by)
            loss.backward()
            optimizer.step()
        scheduler.step()

        # Validation check
        model.eval()
        with torch.no_grad():
            val_preds = model(X_tab_val_t, X_vis_val_t)
            val_loss = criterion(val_preds, y_val_t).item()
            if val_loss < best_val_loss:
                best_val_loss = val_loss
                best_weights = {k: v.cpu().clone() for k, v in model.state_dict().items()}

    # Load best validation checkpoint and predict on Test set
    model.load_state_dict({k: v.to(device) for k, v in best_weights.items()})
    model.eval()
    with torch.no_grad():
        X_tab_te_t = torch.tensor(X_tab_te, dtype=torch.float32).to(device)
        X_vis_te_t = torch.tensor(X_vis_te, dtype=torch.float32).to(device)
        test_preds = model(X_tab_te_t, X_vis_te_t).squeeze(1).cpu().numpy()

    return test_preds, model


# -------------------------------------------------------------------------
# 5. Main Execution Pipeline
# -------------------------------------------------------------------------
def run_phase3_modeling(pca_components: int = 32):
    df = load_and_prepare_master_table()
    
    # Define Pure Tabular Features vs. OpenCV Hand-Crafted Features
    tabular_cols = [
        "clean_price", "log_price", "catalog_avg_rating", "total_catalog_ratings",
        "bullet_count", "monthly_review_count", "monthly_avg_rating",
        "reviews_lag_1m", "reviews_lag_2m", "rolling_3m_velocity",
        "verified_ratio", "month_sin", "month_cos"
    ]
    opencv_cols = [
        "aspect_ratio", "sharpness_score", "brightness_mean",
        "contrast_std", "white_bg_ratio"
    ]
    target_col = "target_log_demand"

    # Drop any nulls in feature columns
    df = df.drop_nulls(subset=tabular_cols + opencv_cols + ["vision_embedding", target_col])
    train_df, val_df, test_df = temporal_phash_disjoint_split(df)

    # Extract NumPy arrays
    X_tab_tr = train_df.select(tabular_cols).to_numpy()
    X_tab_val = val_df.select(tabular_cols).to_numpy()
    X_tab_te = test_df.select(tabular_cols).to_numpy()

    X_cv_tr = train_df.select(opencv_cols).to_numpy()
    X_cv_val = val_df.select(opencv_cols).to_numpy()
    X_cv_te = test_df.select(opencv_cols).to_numpy()

    # Unpack 576-d vision_embedding List column into 2D NumPy matrix (N, 576)
    E_tr = np.vstack(train_df["vision_embedding"].to_numpy())
    E_val = np.vstack(val_df["vision_embedding"].to_numpy())
    E_te = np.vstack(test_df["vision_embedding"].to_numpy())

    y_tr = train_df[target_col].to_numpy()
    y_val = val_df[target_col].to_numpy()
    y_te = test_df[target_col].to_numpy()

    # Fit PCA strictly on Training Vision Embeddings (Zero Test Leakage)
    print(f"\nFitting PCA ({pca_components} components) on {E_tr.shape[0]} training embeddings...")
    pca = PCA(n_components=pca_components, random_state=42)
    Z_pca_tr = pca.fit_transform(E_tr)
    Z_pca_val = pca.transform(E_val)
    Z_pca_te = pca.transform(E_te)
    explained_var = np.sum(pca.explained_variance_ratio_) * 100
    print(f"   -> {pca_components} PCA components captured {explained_var:.2f}% of visual variance.")

    # Concatenate Tabular + OpenCV + PCA Embeddings for Multi-Modal LightGBM
    pca_col_names = [f"vis_pca_{i}" for i in range(pca_components)]
    fused_feature_names = tabular_cols + opencv_cols + pca_col_names

    X_fused_tr = np.hstack([X_tab_tr, X_cv_tr, Z_pca_tr])
    X_fused_val = np.hstack([X_tab_val, X_cv_val, Z_pca_val])
    X_fused_te = np.hstack([X_tab_te, X_cv_te, Z_pca_te])

    results = []

    # ---------------------------------------------------------------------
    # Model 1: Baseline Tabular-Only LightGBM
    # ---------------------------------------------------------------------
    print("Training Model 1: Baseline Tabular-Only LightGBM...")
    lgb_tab = lgb.LGBMRegressor(
        n_estimators=1000, learning_rate=0.03, num_leaves=31,
        subsample_for_bin=50000, colsample_bytree=0.8, random_state=42, verbose=-1
    )
    lgb_tab.fit(
        X_tab_tr, y_tr,
        eval_set=[(X_tab_val, y_val)],
        callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)]
    )
    preds_tab = lgb_tab.predict(X_tab_te)
    results.append(evaluate_predictions(y_te, preds_tab, "1. Tabular-Only LightGBM"))

    # ---------------------------------------------------------------------
    # Model 2: Multi-Modal Late-Fusion LightGBM (Tabular + OpenCV + 32-d PCA)
    # ---------------------------------------------------------------------
    print("Training Model 2: Multi-Modal Fusion LightGBM (Tabular + OpenCV + PCA)...")
    lgb_fused = lgb.LGBMRegressor(
        n_estimators=1000, learning_rate=0.03, num_leaves=31,
        colsample_bytree=0.7, subsample=0.8, random_state=42, verbose=-1
    )
    lgb_fused.fit(
        X_fused_tr, y_tr,
        eval_set=[(X_fused_val, y_val)],
        feature_name=fused_feature_names,
        callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)]
    )
    preds_lgb_fused = lgb_fused.predict(X_fused_te)
    results.append(evaluate_predictions(y_te, preds_lgb_fused, "2. Multi-Modal LightGBM (PCA Fusion)"))

    # ---------------------------------------------------------------------
    # Model 3: PyTorch Gated Multi-Modal Neural Network (Full 576-d Vision)
    # ---------------------------------------------------------------------
    print("Training Model 3: PyTorch Gated Multi-Modal Fusion Network...")
    scaler = StandardScaler()
    X_tab_cv_tr_scaled = scaler.fit_transform(np.hstack([X_tab_tr, X_cv_tr]))
    X_tab_cv_val_scaled = scaler.transform(np.hstack([X_tab_val, X_cv_val]))
    X_tab_cv_te_scaled = scaler.transform(np.hstack([X_tab_te, X_cv_te]))

    preds_nn, pytorch_model = train_pytorch_fusion(
        X_tab_cv_tr_scaled, E_tr, y_tr,
        X_tab_cv_val_scaled, E_val, y_val,
        X_tab_cv_te_scaled, E_te,
        epochs=18, batch_size=256
    )
    results.append(evaluate_predictions(y_te, preds_nn, "3. PyTorch Gated Fusion Net"))

    # ---------------------------------------------------------------------
    # Model 4: Ensemble Blend (LightGBM Fusion + PyTorch Gated Net)
    # ---------------------------------------------------------------------
    preds_blend = 0.65 * preds_lgb_fused + 0.35 * preds_nn
    results.append(evaluate_predictions(y_te, preds_blend, "4. Hybrid Multi-Modal Ensemble"))

    # Display Leaderboard
    leaderboard = pl.DataFrame(results)
    print("\n================ OUT-OF-TIME TEST LEADERBOARD ================")
    print(leaderboard)

    # Save artifacts for Phase 4/5 (SHAP & Streamlit Simulator)
    Path("models").mkdir(exist_ok=True)
    joblib.dump({
        "lgb_fused": lgb_fused,
        "pca": pca,
        "scaler": scaler,
        "feature_names": fused_feature_names,
        "tabular_cols": tabular_cols,
        "opencv_cols": opencv_cols
    }, "models/multimodal_artifacts.joblib")
    print("\nSaved trained models and PCA transformer to models/multimodal_artifacts.joblib")

if __name__ == "__main__":
    run_phase3_modeling(pca_components=32)
```
