```python
import yfinance as yf
import pandas as pd

# Target a specific stock (e.g., Reliance Industries)
ticker = yf.Ticker("RELIANCE.NS")

# 1. Get numerical price data
df = ticker.history(period="1mo")
df = df[['Open', 'High', 'Low', 'Close', 'Volume']]

# 2. Get qualitative news data safely
news_items = ticker.news

# Diagnostic step: Print the actual keys of the first news item so you can see the API structure
if news_items:
    print(f"API Keys available: {list(news_items[0].keys())}\n")

# Safely extract the title using .get(), falling back to 'content' or 'headline' if 'title' is missing
headlines = []
for article in news_items:
    # Try 'title' first, then common alternatives, default to None if all fail
    raw_title = article.get('title') or article.get('headline') or article.get('content')
    if raw_title:
        headlines.append(raw_title)

print("Recent Headlines:")
for h in headlines[:3]:
    print(f"- {h}")
```

```python
from ollama import chat
from pydantic import BaseModel

# Enforce a strict JSON schema for the model's output
class SentimentResult(BaseModel):
    score: float

def analyze_headline(headline: str) -> float:
    prompt = f"""
    Analyze the financial sentiment of this news headline for the stock. 
    Return a score between -1.0 (highly bearish) and 1.0 (highly bullish). 
    Headline: "{headline}"
    """
    
    response = chat(
        model='llama3.1', # Ensure you have run `ollama run llama3.1` in your terminal
        messages=[{'role': 'user', 'content': prompt}],
        format=SentimentResult.model_json_schema(), # Lock output format
        options={'temperature': 0} # Set to 0 for strict determinism
    )
    
    # Parse the exact float back into Python
    result = SentimentResult.model_validate_json(response.message.content)
    return result.score

# Calculate average daily sentiment
scores = [analyze_headline(h) for h in headlines]
daily_sentiment = sum(scores) / len(scores) if scores else 0.0
print(f"Aggregated Daily Sentiment Score: {daily_sentiment:.2f}")
```

```python
# 1. Append the newly calculated sentiment score
df['News_Sentiment'] = daily_sentiment 

# 2. Shift the Close price to predict TOMORROW'S price based on TODAY'S news
df['Target_Next_Day'] = df['Close'].shift(-1)
df = df.dropna()

# 3. Your features are now ready for model training
X = df[['Open', 'High', 'Low', 'Volume', 'News_Sentiment']].values
y = df['Target_Next_Day'].values
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error

# Split chronologically: First 80% is the past (training), final 20% is the future (testing)
split_idx = int(len(X) * 0.8)

X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

# Scale the features (Prices, Volume, and Sentiment)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
import numpy as np

class CustomRegressionTree:
    def __init__(self, max_depth=3):
        self.max_depth = max_depth
        self.tree = None

    def fit(self,X,y,depth=0):
        if depth>= self.max_depth or len(y)<2 or len(np.unique(y))==1:
            return np.mean(y)

        best_feat, best_thresh, best_var = None,None,float('inf')

        n_features = X.shape[1]
        features = np.random.choice(n_features, max(1, int(n_features/2)), replace=False)

        for feat in features:
            thresholds = np.unique(X[:,feat])
            for t in thresholds:
                left_mask = X[:,feat]<= t
                right_mask = X[:, feat] > t
                if sum(left_mask) ==0 or sum(right_mask) == 0: continue

                var_left = np.var(y[left_mask]) * sum(left_mask)
                var_right = np.var(y[right_mask]) * sum(right_mask)
                total_var = var_left + var_right

                if total_var < best_var:
                    best_var, best_feat, best_thresh = total_var, feat, t

        if best_feat is None: return np.mean(y)

        left_mask = X[:, best_feat] <= best_thresh
        right_mask = X[:, best_feat] > best_thresh

        return{
                    'feature' : best_feat,
                    'threshold' : best_thresh,
                    'left' : self.fit(X[left_mask], y[left_mask], depth+1),
                    'right' : self.fit(X[right_mask], y[right_mask], depth+1)
        }

    def predict_row(self,row,node):
            if not isinstance(node,dict): return node
            if row[node['feature']] <= node['threshold']:
                return self.predict_row(row, node['left'])
            else:
                return self.predict_row(row, node['right'])

    def predict(self, X):
            return np.array([self.predict_row(row, self.tree) for row in X])

class CustomRandomForest:
    def __init__(self,n_trees=10,max_depth=3):
        self.n_trees = n_trees
        self.max_depth = max_depth
        self.trees = []

    def fit(self,X,y):
        self.trees = []
        for _ in range(self.n_trees):
            tree = CustomRegressionTree(max_depth=self.max_depth)
            indices = np.random.choice(len(X), len(X), replace=True)
            tree.tree = tree.fit(X[indices], y[indices])
            self.trees.append(tree)

    def predict(self,X):
        tree_preds = np.array([tree.predict(X) for tree in self.trees])
        return np.mean(tree_preds, axis=0)
        
```

```python
# Assuming your CustomRandomForest class from earlier is loaded in memory
print("Training Custom Random Forest on Price + Sentiment...")
rf_model = CustomRandomForest(n_trees=1000, max_depth=100)

rf_model.fit(X_train_scaled, y_train)
print("Training Complete.")
```

```python
# Generate forward-looking predictions
predictions = rf_model.predict(X_test_scaled)

# Calculate the error
mae = mean_absolute_error(y_test, predictions)

# Print a side-by-side comparison of the final test days
print(f"Forward-Looking MAE: ₹{mae:.2f}\n")
print("--- Last 5 Days Simulation ---")
for i in range(1, 6):
    actual = y_test[-i]
    pred = predictions[-i]
    sentiment = X_test[-i][-1] # The unscaled sentiment score
    print(f"Day -{i} | Sentiment: {sentiment:5.2f} | Predicted: ₹{pred:.2f} | Actual: ₹{actual:.2f}")
```
